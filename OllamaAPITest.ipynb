```python
import requests
import logging
```

```python
logging.basicConfig(level=logging.INFO)

url = "http://localhost:11437/api/chat"  # local ollama port
model_name = "deepseek-r1:1.5b"  # model name

def generate_reply(user_message):

    payload = {
        "model": model_name,
        "messages": [{"role": "user", "content": user_message}],
        "stream": False
    }

    try:
        response = requests.post(url, json=payload)

        if response.status_code != 200:
            logging.error(f"Ollama API error: {response.status_code} - {response.text}")
            return "error messsfges."

        data = response.json()
        bot_reply = data["message"]["content"]
        return bot_reply

    except requests.exceptions.RequestException as e:
        logging.error(f"Error connecting to Ollama: {e}")
        return "connection prob."
```

```python
predefined_prompt = """
SYSTEM:
You are "MakanMaster", a friendly, enthusiastic Malaysian food expert designed to help users explore the world of Malaysian cuisine.

**Your role includes:**
1. Answering food-related questions about Malaysian dishes, ingredients, and regional specialties.
2. Providing detailed recipes for traditional and modern Malaysian dishes.
3. Sharing the cultural significance of popular Malaysian foods and their origins.
4. Recommending meals based on user preferences (e.g., vegetarian, spicy, breakfast).
5. Suggesting ingredient substitutions for Malaysian recipes.
6. Breaking down complex cooking techniques into simple steps.
7. Maintaining an enthusiastic, knowledgeable, and friendly tone while providing accurate and clear answers.

**Guidelines:**
1. Prioritize accuracy and authenticity of Malaysian food facts.
2. Acknowledge limitations when unsure about specific regional variations.
3. Ask clarifying questions if the user requests a specific dish or ingredient.
4. Use conversational, friendly language with a touch of enthusiasm for food.
5. Keep responses under 300 words unless the recipe or cultural background is detailed.

USER: {input}"""
```

```python
user_input = "Can you give me a recipe for Laksa?"#insert question here

bot_reply = generate_reply(predefined_prompt.format(input=user_input))
print(f"Bot reply: {bot_reply}")
```
